# Step 1. EDA

# 1. Date Coloumn Handle

```python
df.loc[:, 'Order Date'] = pd.to_datetime(df['Order Date'], format='%d/%m/%Y')
df.loc[:, 'Ship Date'] = pd.to_datetime(df['Ship Date'], format='%d/%m/%Y')
```

```python
print(df['Order Date'].dtype)
print(df['Ship Date'].dtype)
```

# 2. Monthaly Wise Sales

```python
month_wise_sales = df.groupby(df['Order Date'].dt.to_period('M'))['Sales'].sum()
print(month_wise_sales.head(5))
```

```python
month_wise_sales.plot(kind='line', figsize=(12,6), marker='o', title="Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.show()
```

# 2. Yearly Sales 

```python
yearly_sales = df.groupby(df['Order Date'].dt.year)['Sales'].sum()
print(yearly_sales)
```

```python
yearly_sales = df.groupby(df['Order Date'].dt.year)['Sales'].sum()
yearly_sales.plot(kind='bar', figsize=(8,5), title="Yearly Sales Trend")
plt.xlabel("Year")
plt.ylabel("Total Sales")
plt.show()
```

# 3. Category/Sub‑Category Analysis

```python
category_sales = df.groupby('Category')['Sales'].sum()
sub_category_sales = df.groupby('Sub-Category')['Sales'].sum()
print(category_sales)
print()
print(sub_category_sales.sample(5))
```

```python
category_sales.plot(kind='bar', title="Category-wise Sales")
sub_category_sales.plot(kind='bar', figsize=(12,6), title="Sub-Category-wise Sales")
plt.show()
```

# 4. Region & Segment Insights

```python
region_sales = df.groupby('Region')['Sales'].sum()
segment_sales = df.groupby('Segment')['Sales'].sum()
print(region_sales)
print()
print(segment_sales)
```
